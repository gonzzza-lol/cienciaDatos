```python
import pandas as pd
from statsmodels.stats.weightstats import DescrStatsW

RUTA = "datos/eod_stgo/"
hogares = pd.read_csv(RUTA + "Hogares.csv", sep=";", decimal=",", low_memory=False)
personas = pd.read_csv(RUTA + "personas.csv", sep=";", decimal=",", low_memory=False)
edad = pd.read_csv(RUTA + "Edadpersonas.csv", sep=";", decimal=",", low_memory=False)
```

```python
hogaresLasCondesSuma = (hogares["Comuna"] == "LAS CONDES").sum()
```

```python
PersonasHogares = pd.merge(hogares, personas, on="Hogar", how="inner", suffixes=("_hogar", "_persona"))
personasLasCondesSuma = (PersonasHogares["Comuna"] == "LAS CONDES").sum()
personasLasCondes = PersonasHogares[PersonasHogares["Comuna"] == "LAS CONDES"]
personasLasCondesPond = personasLasCondes["Factor_persona"].sum()
```

```python
hogaresLasCondes = hogares[hogares["Comuna"] == "LAS CONDES"]
hogaresLasCondesPond = hogaresLasCondes["Factor"].sum()
```

```python
PersonasEdades = pd.merge(edad, personas, on="Persona", how="inner")
percentilesEdadMetropolitana = DescrStatsW(
    PersonasEdades["Edad"], weights=PersonasEdades["Factor"]
).quantile([0.25, 0.50, 0.75])

PersonasLasCondesEdades = pd.merge(edad, personasLasCondes, on="Persona", how="inner")
percentilesEdadLasCondes = DescrStatsW(
    PersonasLasCondesEdades["Edad"], weights=PersonasLasCondesEdades["Factor_persona"]
).quantile([0.25, 0.50, 0.75])
print(percentilesEdadMetropolitana)
print(percentilesEdadLasCondes)
```
